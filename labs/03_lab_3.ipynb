```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from sklearn import datasets
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

```python
iris = datasets.load_iris()
X = iris.data
y = iris.target
labels = iris.target_names
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
cov_matrix = np.cov(X_scaled.T)
print("Covariance Matrix:\n", cov_matrix)
```

```python
eigenvalues, eigenvectors = np.linalg.eig(cov_matrix)
print("Eigenvalues:", eigenvalues)
print("Eigenvectors:\n", eigenvectors)
```

```python
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')
colors = ['red', 'green', 'blue']

for i in range(len(colors)):
    ax.scatter(
        X_scaled[y == i, 0], X_scaled[y == i, 1], X_scaled[y == i, 2],
        color=colors[i], label=labels[i]
    )

ax.set_xlabel('Sepal Length')
ax.set_ylabel('Sepal Width')
ax.set_zlabel('Petal Length')
ax.set_title('3D Visualization of Iris Data Before PCA')
plt.legend()
plt.show()
```

```python
U, S, Vt = np.linalg.svd(X_scaled, full_matrices=False)
print("Singular Values:", S)
```

```python
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

explained_variance = pca.explained_variance_ratio_
print(f"Explained Variance by PC1: {explained_variance[0]:.2f}")
print(f"Explained Variance by PC2: {explained_variance[1]:.2f}")
```

```python
plt.figure(figsize=(8, 6))
for i in range(len(colors)):
    plt.scatter(
        X_pca[y == i, 0], X_pca[y == i, 1],
        color=colors[i], label=labels[i]
    )

plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('PCA on Iris Dataset (Dimensionality Reduction)')
plt.legend()
plt.show()
```

```python
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')

for i in range(len(colors)):
    ax.scatter(
        X_scaled[y == i, 0], X_scaled[y == i, 1], X_scaled[y == i, 2],
        color=colors[i], label=labels[i]
    )

for i in range(3):
    ax.quiver(
        0, 0, 0,
        eigenvectors[0, i], eigenvectors[1, i], eigenvectors[2, i],
        color='black', length=2.5, normalize=True
    )

ax.set_xlabel('Sepal Length')
ax.set_ylabel('Sepal Width')
ax.set_zlabel('Petal Length')
ax.set_title('3D Data with Eigenvectors')
plt.legend()
plt.show()
```
