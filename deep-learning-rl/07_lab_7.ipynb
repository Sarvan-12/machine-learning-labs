```python
import numpy as np
import matplotlib.pyplot as plt

import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input
from tensorflow.keras.datasets import cifar10
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.models import Model
from sklearn.metrics import confusion_matrix, classification_report
import seaborn as sns
```

```python
(x_train, y_train), (x_test, y_test) = cifar10.load_data()
print(x_train.shape, x_test.shape, y_train.shape, y_test.shape)
```

```python
IMG_SIZE = 96   # MobileNetV2 prefers >=96, commonly 96 or 128
def preprocess_images(images):
    images_resized = tf.image.resize(images, (IMG_SIZE, IMG_SIZE))
    images_resized = tf.cast(images_resized, tf.float32)
    return preprocess_input(images_resized)

x_train_pp = preprocess_images(x_train).numpy()
x_test_pp  = preprocess_images(x_test).numpy()
```

```python
num_classes = 10
y_train_cat = to_categorical(y_train, num_classes)
y_test_cat  = to_categorical(y_test, num_classes)
```

```python
batch_size = 64
train_ds = tf.data.Dataset.from_tensor_slices((x_train_pp, y_train_cat)).shuffle(5000).batch(batch_size).prefetch(tf.data.AUTOTUNE)
test_ds  = tf.data.Dataset.from_tensor_slices((x_test_pp,  y_test_cat)).batch(batch_size).prefetch(tf.data.AUTOTUNE)
```
